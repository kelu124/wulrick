# echomods Community Map

Interactive world map of the 150+ echomods / un0rick community contributors.

Built with [folium](https://python-visualization.github.io/folium/) on a CartoDB Positron basemap. Markers are clustered for readability and color-coded by geographic region. Click any marker for the contributor's name.

Data source: [kelu124/echomods](https://github.com/kelu124/echomods) — `include/community/` subdirectories.

```python
def region_color(lat, lon):
    """Color-code contributors by geographic region."""
    if lon < -30:                              # Americas
        return '#2196F3' if lat > 10 else '#4CAF50'
    elif lon > 100 and lat < -10:              # Oceania
        return '#FF9800'
    elif lon > 100:                            # East / SE Asia
        return '#9C27B0'
    elif lon > 55:                             # South / Central Asia
        return '#00BCD4'
    elif lat < 10 and lon > -20 and lon < 55:  # Sub-Saharan Africa
        return '#795548'
    elif lat < 37 and lon > -20 and lon < 55:  # Middle East / North Africa
        return '#FF5722'
    else:                                      # Europe
        return '#F44336'


m = folium.Map(
    location=[30, 10],
    zoom_start=2,
    tiles='CartoDB positron',
    min_zoom=1,
    max_zoom=10,
)

cluster = MarkerCluster(
    name='Contributors',
    options={
        'maxClusterRadius': 50,
        'showCoverageOnHover': False,
        'spiderfyOnMaxZoom': True,
    }
).add_to(m)

for c in contributors:
    color = region_color(c['lat'], c['lon'])
    popup_html = (
        '<div style="font-family:\'Helvetica Neue\',sans-serif;'
        'padding:6px 10px;font-size:14px;font-weight:600;color:#333;">'
        f"{c['who']}</div>"
    )
    folium.CircleMarker(
        location=[c['lat'], c['lon']],
        radius=7,
        color='white',
        weight=1.5,
        fill=True,
        fill_color=color,
        fill_opacity=0.85,
        popup=folium.Popup(popup_html, max_width=200),
        tooltip=c['who'],
    ).add_to(cluster)

# ── legend ──────────────────────────────────────────────────────────────────
legend_html = '''
<div style="
    position:fixed;bottom:30px;left:30px;
    background:white;border-radius:10px;
    padding:14px 18px;
    box-shadow:0 2px 12px rgba(0,0,0,0.15);
    font-family:'Helvetica Neue',sans-serif;
    font-size:12px;z-index:1000;min-width:170px;
">
  <div style="font-weight:700;margin-bottom:10px;color:#222;font-size:13px;letter-spacing:.02em;">Region</div>
  <div style="margin:3px 0"><span style="display:inline-block;width:11px;height:11px;border-radius:50%;background:#2196F3;margin-right:7px;vertical-align:middle;"></span>North America</div>
  <div style="margin:3px 0"><span style="display:inline-block;width:11px;height:11px;border-radius:50%;background:#4CAF50;margin-right:7px;vertical-align:middle;"></span>South America</div>
  <div style="margin:3px 0"><span style="display:inline-block;width:11px;height:11px;border-radius:50%;background:#F44336;margin-right:7px;vertical-align:middle;"></span>Europe</div>
  <div style="margin:3px 0"><span style="display:inline-block;width:11px;height:11px;border-radius:50%;background:#FF5722;margin-right:7px;vertical-align:middle;"></span>Middle East / N. Africa</div>
  <div style="margin:3px 0"><span style="display:inline-block;width:11px;height:11px;border-radius:50%;background:#795548;margin-right:7px;vertical-align:middle;"></span>Sub-Saharan Africa</div>
  <div style="margin:3px 0"><span style="display:inline-block;width:11px;height:11px;border-radius:50%;background:#00BCD4;margin-right:7px;vertical-align:middle;"></span>South / Central Asia</div>
  <div style="margin:3px 0"><span style="display:inline-block;width:11px;height:11px;border-radius:50%;background:#9C27B0;margin-right:7px;vertical-align:middle;"></span>East / SE Asia</div>
  <div style="margin:3px 0"><span style="display:inline-block;width:11px;height:11px;border-radius:50%;background:#FF9800;margin-right:7px;vertical-align:middle;"></span>Oceania</div>
</div>
'''

# ── title ───────────────────────────────────────────────────────────────────
title_html = f'''
<div style="
    position:fixed;top:16px;left:50%;transform:translateX(-50%);
    background:white;border-radius:10px;
    padding:9px 22px;
    box-shadow:0 2px 12px rgba(0,0,0,0.15);
    font-family:'Helvetica Neue',sans-serif;
    font-size:16px;font-weight:700;color:#222;
    z-index:1000;white-space:nowrap;
">
    echomods community &mdash; {len(contributors)} contributors worldwide
</div>
'''

m.get_root().html.add_child(folium.Element(legend_html))
m.get_root().html.add_child(folium.Element(title_html))

m
```

```python
# ── regional breakdown ───────────────────────────────────────────────────────
REGION_NAMES = {
    '#2196F3': 'North America',
    '#4CAF50': 'South America',
    '#F44336': 'Europe',
    '#FF5722': 'Middle East / N. Africa',
    '#795548': 'Sub-Saharan Africa',
    '#00BCD4': 'South / Central Asia',
    '#9C27B0': 'East / SE Asia',
    '#FF9800': 'Oceania',
}

counts = Counter(REGION_NAMES[region_color(c['lat'], c['lon'])] for c in contributors)

print(f"Total: {len(contributors)} contributors\n")
print(f"{'Region':<28} {'Count':>5}   Bar")
print('-' * 55)
for region, count in sorted(counts.items(), key=lambda x: -x[1]):
    bar = '\u2588' * count
    print(f"{region:<28} {count:>5}   {bar}")
```
